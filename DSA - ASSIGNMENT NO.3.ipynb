# **Q1) Implement bubble sort with an early-exit flag. Print the array after each pass for the input 5, 1, 4, 2, 8, and confirm your passes match the trace given in the theory section.**

```python
# Bubble sort with early exit

a = [5, 1, 4, 2, 8]

for i in range(len(a) - 1):
    swapped = False

    for j in range(len(a) - 1 - i):
        if a[j] > a[j + 1]:
            # Swap two elements
            a[j], a[j + 1] = a[j + 1], a[j]
            swapped = True

    # Print array after each pass
    print("Pass", i + 1, ":", a)

    # Stop if no swapping happened
    if swapped == False:
        break

print("Sorted array:", a)
```

# **Q2) Implement insertion sort. Print the array after each pass for the input 8, 4, 1, 6, 3, and report the total number of shift operations performed.**

```python
# Insertion sort

a = [8, 4, 1, 6, 3]
shifts = 0

for i in range(1, len(a)):
    key = a[i]
    j = i - 1

    # Shift elements to the right
    while j >= 0 and a[j] > key:
        a[j + 1] = a[j]
        j = j - 1
        shifts = shifts + 1

    # Insert the key
    a[j + 1] = key

    # Print array after each pass
    print("Pass", i, ":", a)

print("Sorted array:", a)
print("Total shift operations:", shifts)
```

# **Q3) Compare the two algorithms across different kinds of input.**
# **a) Generate three arrays of 1000 elements: random, already sorted, and reverse sorted.**
# **b) Run both algorithms on each, counting comparisons.**
# **c) Present the results as a 3 × 2 table and state which input is the best case and which is the worst case for each algorithm.**


```python
import random

# (a) Generate three arrays of 1000 elements

random_array = list(range(1000))
random.shuffle(random_array)

sorted_array = list(range(1000))

reverse_array = list(range(999, -1, -1))


# Bubble sort
def bubble_sort(a):
    comparisons = 0

    for i in range(len(a) - 1):
        swapped = False

        for j in range(len(a) - 1 - i):
            comparisons = comparisons + 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swapped = True

        if swapped == False:
            break

    return comparisons


# Insertion sort
def insertion_sort(a):
    comparisons = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            comparisons = comparisons + 1

            if a[j] > key:
                a[j + 1] = a[j]
                j = j - 1
            else:
                break

        a[j + 1] = key

    return comparisons


# (b) Run both algorithms and count comparisons

b_random = bubble_sort(random_array.copy())
i_random = insertion_sort(random_array.copy())

b_sorted = bubble_sort(sorted_array.copy())
i_sorted = insertion_sort(sorted_array.copy())

b_reverse = bubble_sort(reverse_array.copy())
i_reverse = insertion_sort(reverse_array.copy())


# Results table
print("Input\t\tBubble Sort\tInsertion Sort")
print("Random\t\t", b_random, "\t\t", i_random)
print("Sorted\t\t", b_sorted, "\t\t", i_sorted)
print("Reverse Sorted\t", b_reverse, "\t\t", i_reverse)


# (c) Best and worst cases

print("\nBubble Sort:")
print("Best case: Sorted array")
print("Worst case: Reverse sorted array")

print("\nInsertion Sort:")
print("Best case: Sorted array")
print("Worst case: Reverse sorted array")
```

# **Q4) Demonstrate that insertion sort is stable using a list of at least six (name, marks) tuples containing repeated marks. Then change the comparison in your inner loop from > to >= and show that the output order changes.**

```python
# List with repeated marks
students = [
    ("Amit", 80),
    ("Riya", 70),
    ("Neha", 80),
    ("Rahul", 60),
    ("Pooja", 70),
    ("Karan", 80)
]


# Insertion sort using >
def insertion_sort(students):
    a = students.copy()

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j][1] > key[1]:
            a[j + 1] = a[j]
            j = j - 1

        a[j + 1] = key

    return a


# Insertion sort using >=
def insertion_sort_unstable(students):
    a = students.copy()

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j][1] >= key[1]:
            a[j + 1] = a[j]
            j = j - 1

        a[j + 1] = key

    return a


# Stable sorting
print("Using > :")
print(insertion_sort(students))


# Unstable sorting
print("\nUsing >= :")
print(insertion_sort_unstable(students))


# Explanation
print("\nUsing >, students with same marks keep their original order.")
print("Using >=, students with same marks can change their order.")
```

# **Q5) Measure both algorithms at n = 500, 1000 and 2000 on random input, and tabulate the time taken. State which was faster at every size, and whether doubling n roughly quadrupled the time as O(n²) predicts.**

```python
import random
import time

# Bubble sort
def bubble_sort(a):
    for i in range(len(a) - 1):
        for j in range(len(a) - 1 - i):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]


# Insertion sort
def insertion_sort(a):
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j = j - 1

        a[j + 1] = key


# Test for 500, 1000 and 2000 elements
for n in [500, 1000, 2000]:

    # Create random list
    a = list(range(n))
    random.shuffle(a)

    # Bubble sort time
    b = a.copy()
    start = time.time()
    bubble_sort(b)
    bubble = time.time() - start

    # Insertion sort time
    c = a.copy()
    start = time.time()
    insertion_sort(c)
    insertion = time.time() - start

    print("n =", n)
    print("Bubble Sort Time:", round(bubble, 6))
    print("Insertion Sort Time:", round(insertion, 6))
    print()


print("Both algorithms have O(n^2) complexity.")
print("When n is doubled, the time can become about 4 times.")
```
