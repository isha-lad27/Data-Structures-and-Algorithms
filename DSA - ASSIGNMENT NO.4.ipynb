# **Q3) Show that quick sort’s performance depends on the input.**
# **a) Sort an already-sorted array of 500 elements with a last-element pivot, and record the recursion depth reached.**
# **b) Sort a random array of the same size and record the depth again.**
# **c) Explain in two lines why the two depths differ so sharply.**

```python
import random

# Quick sort
def quick_sort(a, low, high, depth):
    if low >= high:
        return depth

    pivot = a[high]
    i = low

    # Partition
    for j in range(low, high):
        if a[j] < pivot:
            a[i], a[j] = a[j], a[i]
            i += 1

    a[i], a[high] = a[high], a[i]

    # Find maximum depth
    left = quick_sort(a, low, i - 1, depth + 1)
    right = quick_sort(a, i + 1, high, depth + 1)

    return max(left, right)


# (a) Already sorted array
a = list(range(500))
d1 = quick_sort(a, 0, 499, 1)

print("(a) Sorted array depth:", d1)


# (b) Random array
b = list(range(500))
random.shuffle(b)

d2 = quick_sort(b, 0, 499, 1)

print("(b) Random array depth:", d2)


# (c) Explanation
print("\n(c) Explanation:")
print("Sorted input gives unbalanced partitions, so recursion becomes deep.")
print("Random input usually gives more balanced partitions, so depth is smaller.")
```

# **Q4) Run bubble sort, insertion sort and quick sort on random arrays of size 500, 1000 and 2000, counting comparisons for each. Present a 3 × 3 table and state which algorithm performed fewest comparisons at every size.**

```python
# Q4: Compare number of comparisons

import random


# Bubble sort
def bubble(a):
    count = 0

    for i in range(len(a) - 1):
        for j in range(len(a) - 1 - i):
            count += 1

            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]

    return count


# Insertion sort
def insertion(a):
    count = 0

    for i in range(1, len(a)):
        key = a[i]
        j = i - 1

        while j >= 0:
            count += 1

            if a[j] > key:
                a[j + 1] = a[j]
                j -= 1
            else:
                break

        a[j + 1] = key

    return count


# Quick sort
def quick(a, low, high):
    if low >= high:
        return 0

    pivot = a[high]
    i = low
    count = 0

    # Partition
    for j in range(low, high):
        count += 1

        if a[j] < pivot:
            a[i], a[j] = a[j], a[i]
            i += 1

    a[i], a[high] = a[high], a[i]

    # Sort both parts
    count += quick(a, low, i - 1)
    count += quick(a, i + 1, high)

    return count


# 3 x 3 comparison table
print("n\tBubble\tInsertion\tQuick")

for n in [500, 1000, 2000]:

    # Create random array
    a = list(range(n))
    random.shuffle(a)

    # Count comparisons
    b = bubble(a.copy())
    i = insertion(a.copy())
    q = quick(a.copy(), 0, n - 1)

    print(n, "\t", b, "\t", i, "\t\t", q)


# Conclusion
print("\nQuick sort usually has the fewest comparisons for random input.")
```

# **Q5) Construct an input that forces your quick sort into its O(n²) worst case, and demonstrate it by measuring the time for that input against a random input of the same size. Explain why that particular input is the worst case for a last-element pivot.**

```python
import random
import time

# Quick sort using last element as pivot
def quick_sort(a, low, high):
    if low >= high:
        return

    pivot = a[high]
    i = low

    for j in range(low, high):
        if a[j] < pivot:
            a[i], a[j] = a[j], a[i]
            i += 1

    a[i], a[high] = a[high], a[i]

    quick_sort(a, low, i - 1)
    quick_sort(a, i + 1, high)


# Input size
n = 2000

# Worst-case input: already sorted
worst = list(range(n))

# Random input
random_list = list(range(n))
random.shuffle(random_list)


# Time worst-case input
start = time.time()
quick_sort(worst, 0, n - 1)
worst_time = time.time() - start


# Time random input
start = time.time()
quick_sort(random_list, 0, n - 1)
random_time = time.time() - start


print("Worst-case time:", round(worst_time, 6))
print("Random input time:", round(random_time, 6))

print("\nWorst case: Already sorted array.")
print("The last element is always the largest, so the partitions are very unbalanced.")
print("Therefore, quick sort takes O(n^2) time.")
```
