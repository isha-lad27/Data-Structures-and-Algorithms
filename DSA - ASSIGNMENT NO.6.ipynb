# **Q1) Implement a hash table using separate chaining, supporting the operations below. Insert the keys 12, 22, 32, 15, 25 and 35 into a table of size 10, print the contents of every bucket, and report the longest chain length.**
# **a) insert(key, value)**
# **b) search(key), returning the value or None**
# **c) delete(key), returning whether the key was present**


```python
# Hash table using separate chaining
size = 10
table = [[] for i in range(size)]


# Insert key and value
def insert(key, value):
    index = key % size
    table[index].append((key, value))


# Search for a key
def search(key):
    index = key % size

    for k, v in table[index]:
        if k == key:
            return v

    return None


# Delete a key
def delete(key):
    index = key % size

    for item in table[index]:
        if item[0] == key:
            table[index].remove(item)
            return True

    return False


# Insert given keys
for key in [12, 22, 32, 15, 25, 35]:
    insert(key, "Value" + str(key))


# Print all buckets
for i in range(size):
    print(i, ":", table[i])


# Find longest chain
longest = max(len(x) for x in table)
print("Longest chain:", longest)


# Test search and delete
print("Search 22:", search(22))
print("Delete 22:", delete(22))
```

# **Q2) Implement insert and search for a hash table that resolves collisions by linear probing. Insert 23, 43, 13, 27 and 33 into a table of size 10, print the slot each occupies, and explain in two lines what problem would arise if a key were later deleted by simply emptying its slot.**


```python
# Hash table using linear probing
size = 10
table = [None] * size


# Insert a key
def insert(key):
    index = key % size

    # Find an empty slot
    while table[index] is not None:
        index = (index + 1) % size

    table[index] = key
    print(key, "-> slot", index)


# Search for a key
def search(key):
    index = key % size
    start = index

    while table[index] is not None:
        if table[index] == key:
            return index

        index = (index + 1) % size

        if index == start:
            break

    return -1


# Insert given keys
for key in [23, 43, 13, 27, 33]:
    insert(key)


print("Table:", table)

# Test search
print("Search 43:", search(43))

# Problem with simple deletion:
print("Simple deletion can break the search chain.")
print("A later key may become unreachable if its slot is emptied.")
```

# **Q3) Show that the table size matters as much as the hash function. Hash the keys 10, 20, 30, 40, 50 and 60 with h(k) = k mod 10 and again with h(k) = k mod 11. Report the number of collisions in each case and explain why a prime table size is preferred.**

```python
# Compare table sizes
keys = [10, 20, 30, 40, 50, 60]


# Hash using mod 10
places10 = []
for key in keys:
    places10.append(key % 10)

print("Using mod 10:", places10)

# Count collisions
print("Collisions with size 10:", len(keys) - len(set(places10)))


# Hash using mod 11
places11 = []
for key in keys:
    places11.append(key % 11)

print("Using mod 11:", places11)

# Count collisions
print("Collisions with size 11:", len(keys) - len(set(places11)))


# Explanation
print("A prime table size usually spreads keys more evenly.")
print("It reduces the chance of repeated hash positions.")
```

# **Q4) Insert 30 randomly chosen keys into separate-chaining tables of size 10, 20 and 50. For each table report the longest chain and the average chain length, and state how the table size affects the cost of a search.**

```python
# Effect of table size

import random

keys = random.sample(range(1, 1000), 30)


# Create hash table
def make_table(size):
    table = [[] for i in range(size)]

    for key in keys:
        index = key % size
        table[index].append(key)

    return table


# Test different table sizes
for size in [10, 20, 50]:

    table = make_table(size)

    longest = max(len(x) for x in table)
    average = len(keys) / size

    print("\nTable size:", size)
    print("Longest chain:", longest)
    print("Average chain length:", round(average, 2))


# Conclusion
print("\nLarger table size usually gives shorter chains.")
print("Shorter chains make searching faster.")
```

# **Q5) Count how often each word appears in a short paragraph, using your own chaining hash table with the string hash from the theory section rather than a Python dictionary. Print the three most frequent words.**

```python
# Count word frequency using chaining

text = "data science is useful and data science is interesting and useful"

# Create hash table
size = 10
table = [[] for i in range(size)]


# String hash function
def hash_word(word):
    total = 0

    for ch in word:
        total += ord(ch)

    return total % size


# Add word to hash table
def add_word(word):
    index = hash_word(word)

    for item in table[index]:
        if item[0] == word:
            item[1] += 1
            return

    table[index].append([word, 1])


# Count all words
words = text.lower().split()

for word in words:
    add_word(word)


# Get all word counts
result = []

for bucket in table:
    for item in bucket:
        result.append(item)


# Sort by frequency
result.sort(key=lambda x: x[1], reverse=True)

# Print three most frequent words
print("Three most frequent words:")

for item in result[:3]:
    print(item[0], ":", item[1])
```
