# **Q1) Implement the following three functions recursively, stating the base case of each.**
# **a) The sum of the first n natural numbers.**
# **b) The factorial of n.**
# **c) The sum of all elements in a list.**


```python
# a) Sum of first n natural numbers
def sum_n(n):
    # Base case
    if n == 0:
        return 0
    return n + sum_n(n - 1)


# b) Factorial of n
def factorial(n):
    # Base case
    if n == 0:
        return 1
    return n * factorial(n - 1)


# c) Sum of all elements in a list
def list_sum(a):
    # Base case
    if len(a) == 0:
        return 0
    return a[0] + list_sum(a[1:])


# Examples
print("Sum of first 5 natural numbers:", sum_n(5))
print("Factorial of 5:", factorial(5))
print("Sum of list:", list_sum([10, 20, 30, 40]))
```

# **Q2) Write a recursive function that counts the number of digits in a positive integer, and another that reverses a string. Test both on at least five inputs, including the smallest valid input for each.**


```python
# Count number of digits in a positive integer
def count_digits(n):
    # Base case
    if n < 10:
        return 1
    return 1 + count_digits(n // 10)


# Reverse a string
def reverse_string(s):
    # Base case
    if len(s) == 0:
        return ""
    return reverse_string(s[1:]) + s[0]


# Test count_digits with 5 inputs
print("Number of digits:")

for n in [1, 5, 12, 123, 12345]:
    print(n, "->", count_digits(n))


# Test reverse_string with 5 inputs
print("\nReverse of strings:")

for s in ["a", "hi", "hello", "python", "student"]:
    print(s, "->", reverse_string(s))


# Smallest valid inputs:
# Positive integer = 1
# String = "a"
```

# **Q3) Compute the nth Fibonacci number recursively, with a counter for the number of calls made. Report the call count for n = 5, 10 and 15 in a table, and explain, using the recursion tree, why the count grows so quickly.**

```python
# Fibonacci function with call counter
def fibonacci(n, count):
    # Count each function call
    count[0] = count[0] + 1

    # Base case
    if n <= 1:
        return n

    # Recursive calls
    return fibonacci(n - 1, count) + fibonacci(n - 2, count)


# Test for n = 5, 10 and 15
values = [5, 10, 15]

print("n\tFibonacci\tCall Count")

for n in values:
    count = [0]

    answer = fibonacci(n, count)

    print(n, "\t", answer, "\t\t", count[0])


# Explanation
print("\nExplanation:")
print("Each call creates two more recursive calls.")
print("So the recursion tree grows very quickly.")
print("The time complexity is O(2^n).")
```

# **Q4) Write a recursive function that counts down from n to 0. Find by trial the largest n it can handle before Python raises RecursionError, report that value, explain in one line what is running out, and rewrite the same countdown iteratively so that no limit is reached.**


```python
# Recursive countdown
def countdown(n):
    if n == 0:
        return
    countdown(n - 1)


# Find largest n before RecursionError
n = 1

while True:
    try:
        countdown(n)
        n = n + 1
    except RecursionError:
        break

print("Largest n handled:", n - 1)
print("RecursionError occurs at:", n)

# Reason
print("Reason: Python recursion call stack limit is reached.")


# Iterative countdown
def countdown_loop(n):
    while n >= 0:
        n = n - 1


# Test iterative version
countdown_loop(1000000)

print("Iterative countdown completed.")
print("No recursion limit is reached.")
```

# **Q5) Compute x raised to the power n in two recursive ways: by multiplying x by itself n times, and by halving the exponent as shown in the theory section. Count the multiplications each performs for n = 16 and n = 100, and explain the difference.**

```python
# Power using simple recursion
def power_simple(x, n, count):
    if n == 0:
        return 1

    count[0] = count[0] + 1
    return x * power_simple(x, n - 1, count)


# Power using exponent halving
def power_fast(x, n, count):
    if n == 0:
        return 1

    if n % 2 == 0:
        half = power_fast(x, n // 2, count)
        count[0] = count[0] + 1
        return half * half
    else:
        count[0] = count[0] + 1
        return x * power_fast(x, n - 1, count)


# Test for n = 16
count1 = [0]
count2 = [0]

print("For n = 16")
print("Simple result:", power_simple(2, 16, count1))
print("Simple multiplications:", count1[0])

print("Fast result:", power_fast(2, 16, count2))
print("Fast multiplications:", count2[0])


# Test for n = 100
count1 = [0]
count2 = [0]

print("\nFor n = 100")
print("Simple result:", power_simple(2, 100, count1))
print("Simple multiplications:", count1[0])

print("Fast result:", power_fast(2, 100, count2))
print("Fast multiplications:", count2[0])


# Explanation
print("\nExplanation:")
print("Simple method takes O(n) multiplications.")
print("Fast method takes O(log n) multiplications.")
print("The fast method is better for large values of n.")
```
