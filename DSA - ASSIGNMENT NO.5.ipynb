# **Q1) Implement linear search and binary search on the same sorted list of 1000 elements. Count the comparisons each makes for a key at the start, a key in the middle, a key at the end, and a key that is absent, and present the results in a table.**

```python
a = list(range(1000))

# Linear search
def linear(key):
    count = 0

    for i in range(1000):
        count += 1
        if a[i] == key:
            return count

    return count


# Binary search
def binary(key):
    low = 0
    high = 999
    count = 0

    while low <= high:
        mid = (low + high) // 2
        count += 1

        if a[mid] == key:
            return count
        elif a[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return count


# Start, middle, end and absent
print("Key\tLinear\tBinary")

for key in [0, 500, 999, 2000]:
    print(key, "\t", linear(key), "\t", binary(key))
```

# **Q2) Implement binary search both iteratively and recursively. Verify that both return the same index for 100 keys drawn from a sorted list, including some keys that are absent.**

```python
a = list(range(1000))


# Iterative binary search
def binary_loop(key):
    low = 0
    high = 999

    while low <= high:
        mid = (low + high) // 2

        if a[mid] == key:
            return mid
        elif a[mid] < key:
            low = mid + 1
        else:
            high = mid - 1

    return -1


# Recursive binary search
def binary_rec(key, low, high):
    if low > high:
        return -1

    mid = (low + high) // 2

    if a[mid] == key:
        return mid
    elif a[mid] < key:
        return binary_rec(key, mid + 1, high)
    else:
        return binary_rec(key, low, mid - 1)


# 100 keys including absent keys
keys = list(range(80)) + list(range(1000, 1020))

same = True

for key in keys:
    x = binary_loop(key)
    y = binary_rec(key, 0, 999)

    if x != y:
        same = False

print("Both methods give same index:", same)
```

# **Q3) Modify binary search to return the number of comparisons it made. Use it to find the largest number of comparisons needed for any key in a sorted list of 1000 elements, and compare that figure with log₂(1000).**

```python
# Sorted list of 1000 elements
a = list(range(1000))


# Binary search function
def binary_search(key):
    low = 0
    high = len(a) - 1
    count = 0

    # Continue until search range becomes empty
    while low <= high:

        # Find middle element
        mid = (low + high) // 2
        count += 1

        # Check if key is found
        if a[mid] == key:
            return count

        # Search in right half
        elif a[mid] < key:
            low = mid + 1

        # Search in left half
        else:
            high = mid - 1

    return count


# Find maximum comparisons
largest = 0

for key in a:
    c = binary_search(key)

    if c > largest:
        largest = c


# Display result
print("Largest comparisons:", largest)
print("log2(1000):", 1000 .bit_length() - 1)
```

# **Q4) Determine when sorting first is worth the cost.**
# **a) Time a single linear search on an unsorted list of 5000 elements.**
# **b) Time the cost of sorting that list and then performing one binary search.**
# **c) Repeat the comparison for 1000 searches on the same list, and state with your own numbers when sorting first pays for itself.**

```python
import random
import time


# Create a list of 5000 numbers
a = list(range(5000))
random.shuffle(a)

# Key to search
key = 2500


# (a) Time one linear search
start = time.time()

for x in a:
    if x == key:
        break

linear_time = time.time() - start


# (b) Sort the list
b = a.copy()

start = time.time()
b.sort()
sort_time = time.time() - start


# Perform one binary search
start = time.time()

low = 0
high = len(b) - 1

while low <= high:
    mid = (low + high) // 2

    if b[mid] == key:
        break

    elif b[mid] < key:
        low = mid + 1

    else:
        high = mid - 1

binary_time = time.time() - start

# Total time for sorting + one search
one_search = sort_time + binary_time


# (c) Perform 1000 linear searches
start = time.time()

for k in range(1000):
    for x in a:
        if x == key:
            break

many_linear = time.time() - start


# Perform 1000 binary searches
start = time.time()

for k in range(1000):
    low = 0
    high = len(b) - 1

    while low <= high:
        mid = (low + high) // 2

        if b[mid] == key:
            break

        elif b[mid] < key:
            low = mid + 1

        else:
            high = mid - 1

many_binary = time.time() - start

# Add sorting time
total_binary = sort_time + many_binary


# Display results
print("One linear search:", round(linear_time, 6))
print("Sort + one binary search:", round(one_search, 6))

print("\n1000 linear searches:", round(many_linear, 6))
print("Sort + 1000 binary searches:", round(total_binary, 6))
```

# **Q5) Apply binary search to a sorted list of at least twenty student names, looking up a name and reporting the number of comparisons used. Compare that count with a linear search for the same name, and state what must be true of the list before binary search can be used at all.**

```python
# Compare linear and binary search

names = [
    "Aarav", "Aditya", "Akash", "Aman", "Ananya",
    "Arjun", "Diya", "Isha", "Kabir", "Karan",
    "Kavya", "Meera", "Neha", "Nikhil", "Pooja",
    "Rahul", "Riya", "Rohan", "Sneha", "Yash"
]

key = input("Enter student name: ")


# Linear search
count1 = 0

for name in names:
    count1 += 1
    if name == key:
        break


# Binary search
low = 0
high = len(names) - 1
count2 = 0

while low <= high:
    mid = (low + high) // 2
    count2 += 1

    if names[mid] == key:
        break
    elif names[mid] < key:
        low = mid + 1
    else:
        high = mid - 1


print("Linear search comparisons:", count1)
print("Binary search comparisons:", count2)

print("The list must be sorted before using binary search.")
```
