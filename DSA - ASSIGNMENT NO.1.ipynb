# **Q1) Write a program that finds the largest and the smallest value in a list of numbers without using max() or min(). Report how many comparisons your code performed and state its time complexity in Big-O notation.**

```python
# List of numbers
a = [10, 5, 25, 3, 18]

# Assume first value is largest and smallest
largest = a[0]
smallest = a[0]
count = 0

# Check all remaining values
for i in range(1, len(a)):
    count += 1
    if a[i] > largest:
        largest = a[i]

    count += 1
    if a[i] < smallest:
        smallest = a[i]

# Print result
print("Largest =", largest)
print("Smallest =", smallest)
print("Comparisons =", count)
print("Time Complexity = O(n)")
```

# **Q3) Investigate how running time grows with input size.**
# **a) Write three functions on a list of size n: one O(1), one O(n) and one O(n²).**
# **b) Time each at n = 500, 1000 and 2000, and present the results as a table of n against time.**
# **c) State which function grew fastest, and whether that matches its Big-O complexity.**


```python
import time

# O(1) function
def constant_time(a):
    return a[0]


# O(n) function
def linear_time(a):
    total = 0
    for x in a:
        total = total + x
    return total


# O(n^2) function
def quadratic_time(a):
    count = 0
    for i in a:
        for j in a:
            count = count + 1
    return count


# Different input sizes
sizes = [500, 1000, 2000]

print("n\tO(1)\t\tO(n)\t\tO(n^2)")

for n in sizes:
    a = list(range(n))

    # Time O(1)
    start = time.time()
    constant_time(a)
    t1 = time.time() - start

    # Time O(n)
    start = time.time()
    linear_time(a)
    t2 = time.time() - start

    # Time O(n^2)
    start = time.time()
    quadratic_time(a)
    t3 = time.time() - start

    print(n, "\t", round(t1, 6), "\t", round(t2, 6), "\t", round(t3, 6))

# Conclusion
print("\nO(n^2) function grew fastest.")
print("This matches its Big-O complexity because O(n^2) grows faster as n increases.")
```

# **Q4) Count the comparisons a linear search makes on a list of 1000 elements for a key that is the first element, a key in the middle, and a key that is absent. Identify which of the three is the best case and which is the worst case, and explain why**

```python
# Linear search function
def linear_search(a, key):
    count = 0

    for i in range(len(a)):
        count = count + 1

        if a[i] == key:
            return count

    return count


# List of 1000 elements
a = list(range(1000))

# Key is the first element
first = linear_search(a, 0)

# Key is the middle element
middle = linear_search(a, 500)

# Key is absent
absent = linear_search(a, 2000)

# Print comparisons
print("First element comparisons:", first)
print("Middle element comparisons:", middle)
print("Absent element comparisons:", absent)

# Best case
print("\nBest case: First element - 1 comparison.")
print("The key is found immediately.")

# Middle case
print("\nMiddle case: 501 comparisons.")
print("The key is found after checking half of the list.")

# Worst case
print("\nWorst case: Absent element - 1000 comparisons.")
print("All elements have to be checked.")

# Time complexity
print("\nTime Complexity: O(n)")
```

# **Q5) Compute the sum 1 + 2 + … + n in two ways: with a loop, and with the formula n × (n + 1) / 2. Verify that both give the same answer for n = 1 to 100, time both at n = 1000000, and state the complexity of each. Explain what this shows about the value of choosing a better algorithm.**

```python
import time

# Sum using loop
def sum_loop(n):
    total = 0
    for i in range(1, n + 1):
        total = total + i
    return total


# Sum using formula
def sum_formula(n):
    return n * (n + 1) // 2


# Check both methods for n = 1 to 100
same = True

for n in range(1, 101):
    if sum_loop(n) != sum_formula(n):
        same = False

print("Both methods give same answer:", same)


# Time both methods for n = 1000000
n = 1000000

start = time.time()
sum_loop(n)
loop_time = time.time() - start

start = time.time()
sum_formula(n)
formula_time = time.time() - start

print("Loop time:", loop_time)
print("Formula time:", formula_time)

print("Loop complexity: O(n)")
print("Formula complexity: O(1)")


# Conclusion
# Both methods give the same answer.
# The loop takes O(n) time because it adds every number one by one.
# The formula takes O(1) time because it calculates the answer directly.
# This shows that choosing a better algorithm can save time for large inputs.
```
