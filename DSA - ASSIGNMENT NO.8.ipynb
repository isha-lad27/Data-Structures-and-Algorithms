# **Q1) Implement a doubly linked list supporting the operations listed below, with a test driver covering an empty list, a one-element list and deletion of the only node.**
# **a) insert_at_head, insert_at_tail and insert_after a given node**
# **b) delete_by_value, correctly handling head, tail and middle nodes**
# **c) forward traversal and backward traversal**

```python
# Doubly linked list

class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DLL:
    def __init__(self):
        self.head = None
        self.tail = None

    # Insert at beginning
    def insert_head(self, data):
        n = Node(data)

        if self.head is None:
            self.head = self.tail = n
        else:
            n.next = self.head
            self.head.prev = n
            self.head = n

    # Insert at end
    def insert_tail(self, data):
        n = Node(data)

        if self.tail is None:
            self.head = self.tail = n
        else:
            n.prev = self.tail
            self.tail.next = n
            self.tail = n

    # Insert after a node
    def insert_after(self, node, data):
        n = Node(data)

        n.prev = node
        n.next = node.next

        if node.next:
            node.next.prev = n
        else:
            self.tail = n

        node.next = n

    # Delete a value
    def delete(self, data):
        temp = self.head

        while temp:
            if temp.data == data:

                if temp.prev:
                    temp.prev.next = temp.next
                else:
                    self.head = temp.next

                if temp.next:
                    temp.next.prev = temp.prev
                else:
                    self.tail = temp.prev

                return

            temp = temp.next

    # Print from head to tail
    def forward(self):
        temp = self.head

        while temp:
            print(temp.data, end=" ")
            temp = temp.next
        print()

    # Print from tail to head
    def backward(self):
        temp = self.tail

        while temp:
            print(temp.data, end=" ")
            temp = temp.prev
        print()


# Test
d = DLL()

print("Empty list:")
d.forward()

d.insert_head(10)

print("One element:")
d.forward()
d.backward()

d.insert_tail(20)
d.insert_tail(30)

print("After adding elements:")
d.forward()

# Insert 15 after 10
d.insert_after(d.head, 15)

print("After inserting 15:")
d.forward()

# Delete head
d.delete(10)
print("After deleting head:")
d.forward()

# Delete middle
d.delete(15)
print("After deleting middle:")
d.forward()

# Delete tail
d.delete(30)
print("After deleting tail:")
d.forward()

# Delete only remaining node
d.delete(20)
print("After deleting only node:")
d.forward()
```

# **Q3) Build a doubly linked list of 20 elements. Delete the head, the tail and a middle node, printing the list in both directions after each deletion. Draw the references before and after the middle deletion by hand and attach the drawing.**

```python
# Delete nodes from doubly linked list

class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class DLL:
    def __init__(self):
        self.head = None
        self.tail = None

    # Add node at end
    def add(self, data):
        new = Node(data)

        if self.head is None:
            self.head = self.tail = new
        else:
            new.prev = self.tail
            self.tail.next = new
            self.tail = new

    # Delete a node
    def delete(self, data):
        temp = self.head

        while temp:
            if temp.data == data:

                if temp.prev:
                    temp.prev.next = temp.next
                else:
                    self.head = temp.next

                if temp.next:
                    temp.next.prev = temp.prev
                else:
                    self.tail = temp.prev

                return

            temp = temp.next

    # Print forward
    def forward(self):
        temp = self.head

        while temp:
            print(temp.data, end=" ")
            temp = temp.next
        print()

    # Print backward
    def backward(self):
        temp = self.tail

        while temp:
            print(temp.data, end=" ")
            temp = temp.prev
        print()


# Create 20 elements
d = DLL()

for i in range(1, 21):
    d.add(i)

print("Original list:")
d.forward()
d.backward()


# Delete head
d.delete(1)

print("\nAfter deleting head:")
d.forward()
d.backward()


# Delete tail
d.delete(20)

print("\nAfter deleting tail:")
d.forward()
d.backward()


# Delete middle node
d.delete(10)

print("\nAfter deleting middle node:")
d.forward()
d.backward()
```

# **Q4) Use your doubly linked list to implement a simple browser history supporting visit(page), back() and forward(). Demonstrate it with at least eight operations, and explain why a doubly linked list suits this problem better than a singly linked one.**

```python
# Browser history using doubly linked list

class Page:
    def __init__(self, name):
        self.name = name
        self.prev = None
        self.next = None


class Browser:
    def __init__(self):
        self.current = None

    # Visit a new page
    def visit(self, name):
        new = Page(name)

        if self.current:
            self.current.next = new
            new.prev = self.current

        self.current = new
        print("Visit:", name)

    # Go back
    def back(self):
        if self.current and self.current.prev:
            self.current = self.current.prev
            print("Back:", self.current.name)
        else:
            print("Cannot go back")

    # Go forward
    def forward(self):
        if self.current and self.current.next:
            self.current = self.current.next
            print("Forward:", self.current.name)
        else:
            print("Cannot go forward")


# Test browser history
b = Browser()

b.visit("Google")
b.visit("YouTube")
b.visit("Wikipedia")
b.visit("GitHub")

b.back()
b.back()

b.forward()

b.visit("Instagram")

b.back()

print("Explanation - A doubly linked list is suitable because every page has a prev and next link. Therefore, we can easily move backward and forward through browser history. A singly linked list mainly supports movement in one direction.")
```

# **Q5) Compare deletion in the two kinds of list. Implement a method that deletes a node you already hold a reference to, for both a singly and a doubly linked list of 5000 elements. Measure both, and explain why one is O(1) while the other is O(n).**

```python
# Compare deletion in singly and doubly linked lists

import time


# Node for singly linked list
class SNode:
    def __init__(self, data):
        self.data = data
        self.next = None


# Node for doubly linked list
class DNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


# Delete from singly linked list
def delete_singly(head, node):
    temp = head

    # Find the node before the given node
    while temp.next != node:
        temp = temp.next

    temp.next = node.next
    return head


# Delete from doubly linked list
def delete_doubly(node):
    if node.prev:
        node.prev.next = node.next

    if node.next:
        node.next.prev = node.prev


# Create singly linked list
head = SNode(0)
temp = head

for i in range(1, 5000):
    temp.next = SNode(i)
    temp = temp.next

# Node to delete
s_node = head

for i in range(2500):
    s_node = s_node.next


# Create doubly linked list
d_head = DNode(0)
temp = d_head

for i in range(1, 5000):
    new = DNode(i)
    temp.next = new
    new.prev = temp
    temp = new

d_node = d_head

for i in range(2500):
    d_node = d_node.next


# Time singly linked list deletion
start = time.time()
head = delete_singly(head, s_node)
s_time = time.time() - start


# Time doubly linked list deletion
start = time.time()
delete_doubly(d_node)
d_time = time.time() - start


# Print results
print("Singly linked list:", round(s_time, 6))
print("Doubly linked list:", round(d_time, 6))

print("\nSingly linked list deletion: O(n)")
print("Doubly linked list deletion: O(1)")

print("\nConclusion:")
print("Singly list must find the previous node first.")
print("Doubly list already has a previous reference, so deletion is direct.")
```
