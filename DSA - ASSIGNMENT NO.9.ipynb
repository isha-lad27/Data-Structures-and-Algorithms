# **Q1) Implement the stack ADT twice and compare the two implementations.**
# **a) Once over a fixed-size array, with overflow and underflow detection.**
# **b) Once over a linked list, exposing exactly the same public interface.**
# **c) Test both with the same sequence of operations, confirm identical results, and state one advantage of each implementation.**


```python
# Stack using array and linked list

# Stack using fixed-size array
class ArrayStack:
    def __init__(self, size):
        self.stack = []
        self.size = size

    def push(self, x):
        if len(self.stack) == self.size:
            print("Overflow")
        else:
            self.stack.append(x)

    def pop(self):
        if len(self.stack) == 0:
            print("Underflow")
            return None
        return self.stack.pop()

    def peek(self):
        if len(self.stack) == 0:
            return None
        return self.stack[-1]


# Stack using linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedStack:
    def __init__(self):
        self.top = None

    def push(self, x):
        new = Node(x)
        new.next = self.top
        self.top = new

    def pop(self):
        if self.top is None:
            print("Underflow")
            return None

        x = self.top.data
        self.top = self.top.next
        return x

    def peek(self):
        if self.top is None:
            return None
        return self.top.data


# Test both stacks
a = ArrayStack(3)
b = LinkedStack()

for x in [10, 20, 30]:
    a.push(x)
    b.push(x)

print("Array peek:", a.peek())
print("Linked peek:", b.peek())

print("Array pop:", a.pop())
print("Linked pop:", b.pop())

print("Array pop:", a.pop())
print("Linked pop:", b.pop())

# Advantages
print("\nArray advantage: simple and uses less extra memory.")
print("Linked list advantage: size can grow when needed.")
```

# **Q2) Check whether an expression has balanced brackets of three kinds, reporting the position of the first mismatch rather than only True or False. Test at least six expressions, including correctly nested, wrongly nested and unclosed cases.**

```python
# Check balanced brackets

def check_brackets(text):
    stack = []
    pairs = {')': '(', ']': '[', '}': '{'}

    for i in range(len(text)):
        ch = text[i]

        # Opening bracket
        if ch in "([{":
            stack.append((ch, i))

        # Closing bracket
        elif ch in ")]}":
            if len(stack) == 0:
                return "Mismatch at position " + str(i)

            top, pos = stack.pop()

            if top != pairs[ch]:
                return "Mismatch at position " + str(i)

    # Check for unclosed bracket
    if len(stack) > 0:
        return "Unclosed bracket at position " + str(stack[-1][1])

    return "Balanced"


# Test expressions
tests = [
    "({[]})",
    "[()]",
    "([{}])",
    "([)]",
    "((())",
    "{[}]"
]

for x in tests:
    print(x, ":", check_brackets(x))
```


# **Q3) Build an expression evaluator using stacks.**
# **a) Convert an infix expression to postfix, supporting + − * / with correct precedence and brackets.**
# **b) Evaluate the resulting postfix expression.**
# **c) Chain the two so the user enters infix and receives a numerical answer. Test at least six expressions, including nested brackets and one malformed input that must be rejected gracefully.**


```python
# Infix to postfix and evaluate

def postfix(expr):
    stack = []
    output = []

    priority = {"+": 1, "-": 1, "*": 2, "/": 2}

    for ch in expr:
        if ch.isdigit():
            output.append(ch)

        elif ch == "(":
            stack.append(ch)

        elif ch == ")":
            while stack and stack[-1] != "(":
                output.append(stack.pop())

            if not stack:
                return None

            stack.pop()

        elif ch in "+-*/":
            while stack and stack[-1] != "(" and priority[stack[-1]] >= priority[ch]:
                output.append(stack.pop())

            stack.append(ch)

        elif ch != " ":
            return None

    while stack:
        if stack[-1] == "(":
            return None
        output.append(stack.pop())

    return output


def calculate(post):
    stack = []

    for x in post:
        if x.isdigit():
            stack.append(int(x))
        else:
            if len(stack) < 2:
                return None

            b = stack.pop()
            a = stack.pop()

            if x == "+":
                stack.append(a + b)
            elif x == "-":
                stack.append(a - b)
            elif x == "*":
                stack.append(a * b)
            elif x == "/":
                if b == 0:
                    return None
                stack.append(a / b)

    if len(stack) != 1:
        return None

    return stack[0]


# Test expressions
tests = [
    "2+3",
    "2+3*4",
    "(2+3)*4",
    "8/(2+2)",
    "2*(3+(4*5))",
    "10-2*3",
    "2+(3*4"       # wrong expression
]

for exp in tests:
    p = postfix(exp)

    if p is None:
        print(exp, "-> Invalid expression")
    else:
        ans = calculate(p)

        if ans is None:
            print(exp, "-> Invalid expression")
        else:
            print(exp, "=", ans)
```

# **Q4) Demonstrate why the circular queue exists.**
# **a) Implement a simple array queue, then fill it, empty it, and show that it still reports itself full.**
# **b) Implement a circular queue that fixes the problem.**
# **c) State in a comment which full-versus-empty convention you chose, and why.**

```python
# Simple queue vs circular queue


# Simple queue
class SimpleQueue:
    def __init__(self, size):
        self.q = [None] * size
        self.size = size
        self.front = 0
        self.rear = -1

    def enqueue(self, x):
        if self.rear == self.size - 1:
            print("Queue is full")
        else:
            self.rear += 1
            self.q[self.rear] = x

    def dequeue(self):
        if self.front > self.rear:
            print("Queue is empty")
            return None

        x = self.q[self.front]
        self.front += 1
        return x


# Circular queue
class CircularQueue:
    def __init__(self, size):
        self.q = [None] * size
        self.size = size
        self.front = 0
        self.rear = 0
        self.count = 0

    def enqueue(self, x):
        if self.count == self.size:
            print("Queue is full")
            return

        self.q[self.rear] = x
        self.rear = (self.rear + 1) % self.size
        self.count += 1

    def dequeue(self):
        if self.count == 0:
            print("Queue is empty")
            return None

        x = self.q[self.front]
        self.front = (self.front + 1) % self.size
        self.count -= 1
        return x


# Test simple queue
print("Simple Queue")

q1 = SimpleQueue(3)

q1.enqueue(10)
q1.enqueue(20)
q1.enqueue(30)

q1.dequeue()
q1.dequeue()
q1.dequeue()

# Space at beginning is free, but rear is already at the end
q1.enqueue(40)


# Test circular queue
print("\nCircular Queue")

q2 = CircularQueue(3)

q2.enqueue(10)
q2.enqueue(20)
q2.enqueue(30)

q2.dequeue()
q2.dequeue()

# Circular queue reuses the empty spaces
q2.enqueue(40)
q2.enqueue(50)

print("Circular queue works correctly.")


# Full/empty convention:
# count == 0 means empty
# count == size means full
# This makes full and empty easy to distinguish.
```

# **Q5) Use your circular queue to simulate round-robin scheduling of four processes with a fixed time quantum. Print the order in which the processes run and the completion time of each. Repeat with quanta of 2 and 4, and state which gave the lower average completion time.**

```python
# Round-robin scheduling using circular queue

class CircularQueue:
    def __init__(self, size):
        self.q = [None] * size
        self.size = size
        self.front = 0
        self.rear = 0
        self.count = 0

    # Add process to queue
    def enqueue(self, x):
        if self.count < self.size:
            self.q[self.rear] = x
            self.rear = (self.rear + 1) % self.size
            self.count += 1

    # Remove process from queue
    def dequeue(self):
        if self.count == 0:
            return None

        x = self.q[self.front]
        self.front = (self.front + 1) % self.size
        self.count -= 1
        return x


# Round-robin scheduling
def round_robin(quantum):
    processes = {
        "P1": 5,
        "P2": 4,
        "P3": 2,
        "P4": 6
    }

    q = CircularQueue(4)

    for p in processes:
        q.enqueue(p)

    remaining = processes.copy()
    finish = {}
    time = 0
    order = []

    while q.count > 0:
        p = q.dequeue()

        run = min(quantum, remaining[p])
        time += run
        remaining[p] -= run
        order.append(p)

        if remaining[p] > 0:
            q.enqueue(p)
        else:
            finish[p] = time

    print("\nQuantum:", quantum)
    print("Order:", order)
    print("Completion times:", finish)

    average = sum(finish.values()) / len(finish)
    print("Average completion time:", average)

    return average


# Test with quantum 2 and 4
avg2 = round_robin(2)
avg4 = round_robin(4)

# Compare results
print("\nComparison:")
if avg2 < avg4:
    print("Quantum 2 gives lower average completion time.")
elif avg4 < avg2:
    print("Quantum 4 gives lower average completion time.")
else:
    print("Both give the same average completion time.")
```
