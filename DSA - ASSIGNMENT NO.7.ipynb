# **Q1) Implement a singly linked list class supporting the operations listed below, together with a test driver that exercises every one of them on an empty list, a single-element list and a longer list.**
# **a) insert_at_head, insert_at_tail and insert_at_position**
# **b) delete_by_value and delete_at_position**
# **c) search, length and display**


```python
# Singly linked list

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Insert at head
    def insert_at_head(self, data):
        new = Node(data)
        new.next = self.head
        self.head = new

    # Insert at tail
    def insert_at_tail(self, data):
        new = Node(data)

        if self.head is None:
            self.head = new
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new

    # Insert at position
    def insert_at_position(self, data, pos):
        if pos == 0:
            self.insert_at_head(data)
            return

        temp = self.head

        for i in range(pos - 1):
            temp = temp.next

        new = Node(data)
        new.next = temp.next
        temp.next = new

    # Delete by value
    def delete_by_value(self, data):
        if self.head is None:
            return

        if self.head.data == data:
            self.head = self.head.next
            return

        temp = self.head

        while temp.next:
            if temp.next.data == data:
                temp.next = temp.next.next
                return
            temp = temp.next

    # Delete at position
    def delete_at_position(self, pos):
        if pos == 0:
            self.head = self.head.next
            return

        temp = self.head

        for i in range(pos - 1):
            temp = temp.next

        temp.next = temp.next.next

    # Search
    def search(self, data):
        temp = self.head
        pos = 0

        while temp:
            if temp.data == data:
                return pos
            temp = temp.next
            pos += 1

        return -1

    # Length
    def length(self):
        count = 0
        temp = self.head

        while temp:
            count += 1
            temp = temp.next

        return count

    # Display
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


# Test
l = LinkedList()

print("Empty list:")
l.display()

l.insert_at_head(10)
print("After inserting 10:")
l.display()

l.insert_at_tail(20)
l.insert_at_tail(30)
l.insert_at_position(15, 1)

print("Longer list:")
l.display()

print("Search 20:", l.search(20))
print("Length:", l.length())

l.delete_by_value(15)
print("After deleting 15:")
l.display()

l.delete_at_position(1)
print("After deleting position 1:")
l.display()
```

# **Q2) Reverse a singly linked list iteratively using the three-reference method shown in the theory section. Verify it on lists of length 0, 1, 2 and 10, and state its time and space complexity.**

```python
# Reverse a singly linked list

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


# Reverse the list
def reverse(head):
    prev = None
    current = head

    while current:
        next_node = current.next
        current.next = prev
        prev = current
        current = next_node

    return prev


# Display list
def display(head):
    while head:
        print(head.data, end=" -> ")
        head = head.next

    print("None")


# Create a list
head = Node(1)
head.next = Node(2)
head.next.next = Node(3)
head.next.next.next = Node(4)

print("Before:")
display(head)

# Reverse list
head = reverse(head)

print("After:")
display(head)

print("Time Complexity: O(n)")
print("Space Complexity: O(1)")
```

# **Q3) Add a method that returns the data stored at a given position, and use it together with your length() method to print the middle element of the list. Test it on lists of both odd and even length, and state the complexity of your method.**

```python
# Find the middle element

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Add node at the end
    def add(self, data):
        new = Node(data)

        if self.head is None:
            self.head = new
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new

    # Find length of list
    def length(self):
        count = 0
        temp = self.head

        while temp:
            count += 1
            temp = temp.next

        return count

    # Get data at a position
    def get(self, pos):
        temp = self.head

        for i in range(pos):
            temp = temp.next

        return temp.data


# Odd length list
a = LinkedList()

for x in [10, 20, 30, 40, 50]:
    a.add(x)

middle = a.length() // 2
print("Odd list:", a.get(middle))


# Even length list
b = LinkedList()

for x in [10, 20, 30, 40, 50, 60]:
    b.add(x)

middle = b.length() // 2
print("Even list:", b.get(middle))


# Complexity
print("Complexity: O(n)")
```

# **Q4) Compare a linked list with a Python list.**
# **a) Build both structures with 10000 elements each.**
# **b) Measure the time to insert 2000 elements at the front of each.**
# **c) Measure the time to read 2000 elements at random positions in each, then state which structure wins each test and explain why.**


```python
# Compare Python list and linked list

import time
import random

n = 10000

# Create Python list
a = list(range(n))


# Create linked list
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


head = Node(0)
temp = head

for i in range(1, n):
    temp.next = Node(i)
    temp = temp.next


# (b) Insert 2000 elements at front

start = time.time()

for i in range(2000):
    a.insert(0, i)

python_insert = time.time() - start


start = time.time()

for i in range(2000):
    new = Node(i)
    new.next = head
    head = new

linked_insert = time.time() - start


# (c) Read 2000 random positions

positions = [random.randint(0, n - 1) for i in range(2000)]


# Read from Python list
start = time.time()

for p in positions:
    x = a[p]

python_read = time.time() - start


# Read from linked list
start = time.time()

for p in positions:
    temp = head

    for i in range(p):
        temp = temp.next

    x = temp.data

linked_read = time.time() - start


# Print results
print("Python list insert:", round(python_insert, 5))
print("Linked list insert:", round(linked_insert, 5))

print("Python list read:", round(python_read, 5))
print("Linked list read:", round(linked_read, 5))


# Conclusion
print("\nConclusion:")
print("Linked list is faster for inserting at the front.")
print("Python list is faster for reading random positions.")
print("Linked list needs to travel through nodes to reach a position.")
```

# **Q5) Write a method that deletes every node holding a given value, not only the first, and a second method that removes duplicate values from a sorted linked list. Verify both, including the case where the value to delete is at the head and the case where every node must be removed.**

```python
# Q5: Delete all values and remove duplicates

class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Add node at the end
    def add(self, data):
        new = Node(data)

        if self.head is None:
            self.head = new
            return

        temp = self.head
        while temp.next:
            temp = temp.next

        temp.next = new

    # Delete every node having the given value
    def delete_all(self, value):
        while self.head and self.head.data == value:
            self.head = self.head.next

        temp = self.head

        while temp and temp.next:
            if temp.next.data == value:
                temp.next = temp.next.next
            else:
                temp = temp.next

    # Remove duplicates from sorted list
    def remove_duplicates(self):
        temp = self.head

        while temp and temp.next:
            if temp.data == temp.next.data:
                temp.next = temp.next.next
            else:
                temp = temp.next

    # Display list
    def display(self):
        temp = self.head

        while temp:
            print(temp.data, end=" -> ")
            temp = temp.next

        print("None")


# Test delete all
a = LinkedList()

for x in [5, 5, 10, 15, 5, 20, 5]:
    a.add(x)

print("Original list:")
a.display()

a.delete_all(5)

print("After deleting all 5:")
a.display()


# Test when every node has the same value
b = LinkedList()

for x in [7, 7, 7, 7]:
    b.add(x)

print("\nAll same values:")
b.display()

b.delete_all(7)

print("After deleting all 7:")
b.display()


# Test remove duplicates
c = LinkedList()

for x in [1, 1, 2, 2, 2, 3, 4, 4]:
    c.add(x)

print("\nSorted list:")
c.display()

c.remove_duplicates()

print("After removing duplicates:")
c.display()


# Conclusion
print("\nConclusion:")
print("delete_all() removes every node with the given value.")
print("remove_duplicates() removes repeated values from a sorted list.")
```
